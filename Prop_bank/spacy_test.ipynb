```python
import requests
from bs4 import BeautifulSoup

def extract_txt_from_url(url):
    response = requests.get(url)
    if response.status_code == 200:
        soup = BeautifulSoup(response.content, 'html.parser')
        for tag in soup(['script', 'style']):
            tag.decompose()

        text = ' '.join(soup.stripped_strings)
        return text
    else:
        print(f"Failed to fetch {url}")
        return None


url = "https://news.iu.edu/live/news/26702-indiana-cities-to-develop-climate-action-plans-in"
text = extract_txt_from_url(url)
print(text)
```

### now url link scraping, text to sentence conversion

```python
import requests
from bs4 import BeautifulSoup
import spacy
```

```python
def scrape_txt_from_url(url):
    response = requests.get(url)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, 'html.parser')
    para = soup.find_all('p')
    text = ' '.join([pa.get_text() for pa in para])

    return text 


def extract_sentences_from_txt(text):
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(text)
    sentences = [sent.text for sent in doc.sents]
    return sentences
```

```python
url = "https://news.iu.edu/live/news/26932-iupui-grand-park-release-results-of-study-on-youth"
scraped_text = scrape_txt_from_url(url)
sentences = extract_sentences_from_txt(scraped_text)

for idx, sentence in enumerate(sentences):
    print(f"Sentence {idx + 1}: {sentence}")

with open("scraped_sentences.txt", "w") as file:
    file.write("\n".join(sentences))
```
